```python
from  matplotlib import pyplot as plt
from matplotlib import style

style.use('ggplot')
x=[2,3,4]
x2=[6,5,4]
x3=[7,8,4]
x4=[8,3,2]

plt.plot(x,x2,label='first',linewidth=3)
plt.plot(x3,x4,label='second')
plt.legend()
plt.show()
```
